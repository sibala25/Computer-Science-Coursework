```python
from scipy.stats import t
import numpy as np
from sklearn.linear_model import LinearRegression

# Define X (feature) and y (target)
X = df3['duration'].to_numpy().reshape(-1,1)  # Ensure X is 2D for regression
y = df3['next'].to_numpy()  # y should remain 1D

# Fit the linear regression model
model5 = LinearRegression()
model5.fit(X, y)

# Number of observations
n = len(X)

# Predict Y-hat (fitted values)
y_hat = model5.predict(X)

# Compute residuals and standard error of estimate (s_e)
residuals = y - y_hat
s_e = np.sqrt(np.sum(residuals**2) / (n - 2))  # Standard error of estimate

# Compute mean of X (duration)
x_mean = np.mean(X)

# Compute t-value for 95% CI
t_value = t.ppf(0.975, df=n-2)

# Define new X-value for prediction (e.g., duration = 5)
x_new = np.array([[5]])  # Ensure correct shape

# Compute standard error of prediction
s_pred = s_e * np.sqrt(1 + (1/n) + ((x_new - x_mean)**2 / np.sum((X - x_mean)**2)))

# Predict Y for X = 5
y_pred = model5.predict(x_new)[0]  # Extract scalar

# Compute prediction interval
pred_interval = [y_pred - t_value * s_pred, y_pred + t_value * s_pred]

# Fix formatting error by converting NumPy arrays to scalars
print(f"Predicted next aftershock time: {y_pred:.2f} minutes")
print(f"95% Prediction Interval: [{float(pred_interval[0]):.2f}, {float(pred_interval[1]):.2f}] minutes")
```
