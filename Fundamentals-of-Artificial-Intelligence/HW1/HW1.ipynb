```python
# Question 1
def PAD(n):
  '''
  A function which given an integer, n, returns the value of the nth element in the Padovan sequence.

  Arguments:
  - n: integer, number element of the sequence to return

  Returns:
  - integer, value of the element in the Padovan sequence
  '''
  # base cases of Padovan sequence
  if n==0 or n==1 or n==2:
    return 1
  # if not a base case, use the formula for the sequence
  else:
    return PAD(n-2)+PAD(n-3)
```

```python
for i in range(0,6):
  print(i,PAD(i))
```

```python
# Question 2
def SUMS(n):
  '''
  A function which given an integer, n, returns the number of additions
  needed by above PAD function to compute nth element in the Padovan sequence.

  Arguments:
  - n: integer, number element of the sequence to return

  Returns:
  - integer, number of additions needed to compute nth Padovan number
  '''

  # base cases of Padovan sequence do not require computation
  # so return 0
  if n==0 or n==1 or n==2:
    return 0
  # if not a base case, each time you compute a sum, add 1
  # to account for that computation
  else:
    return SUMS(n-2)+SUMS(n-3)+1
```

```python
for i in range(0,10):
  print(i,SUMS(i))
```

```python
# Question 3
def ANON(TREE):
  '''
  A function which given a single argument, TREE, returns an anonymized tree,
  with the same structure. The original leaves of the tree are replaced with '?'.

  Arguments:
  - TREE: A tree represented through tuples, where elements of a tuple share a parent

  Returns:
  - tree, same as TREE, but leaves are replaced with '?'
  '''

  # base case, when there is just 1 element replace it with '?'
  if type(TREE) is not tuple:
    return '?'
  # when TREE is still a tuple, go a level deeper through recursion and replace
  # elements with '?' using base case
  else:
    return tuple(ANON(x) for x in TREE)
```

```python
ANON(42)
```

```python
ANON("FOO")
```

```python
ANON(((("L", "E"), "F"), "T"))
```

```python
ANON((5, "FOO", 3.1, -0.2))
```

```python
ANON((1, ("FOO", 3.1), -0.2))
```

```python
ANON((((1, 2), ("FOO", 3.1)), ("BAR", -0.2)))
```

```python
ANON(("R", ("I", ("G", ("H", "T")))))
```

```python
# Question 4
def TREE_HEIGHT(TREE):
  '''
  A function which given a single argument, TREE, returns the height of the tree.
  The height is defined as the length of the longest path from root node to farthest
  leaf node.

  Arguments:
  - TREE: A tree represented through tuples, where elements of a tuple share a parent

  Returns:
  - integer, tree height
  '''

  # base case, when there is just 1 node, so height 0
  if type(TREE) is not tuple:
    return 0
  # when TREE is still a tuple, go a level deeper through recursion and
  # increment height for level
  else:
    return max(TREE_HEIGHT(x) for x in TREE) + 1
```

```python
TREE_HEIGHT(1)
```

```python
TREE_HEIGHT((5, "FOO", 3.1, -0.2))
```

```python
TREE_HEIGHT((1, ("FOO", 3.1), -0.2))
```

```python
TREE_HEIGHT(("R", ("I", ("G", ("H", "T")))))
```

```python
# Question 5
def TREE_ORDER(TREE):
  '''
  A function which given a single ordered tree, TREE, returns a tuple which
  representes the postorder traversal of the tree.

  Arguments:
  - TREE: An ordered tree represented through tuples, where elements of a tuple share a parent

  Returns:
  - tuple, of the postorder traversal of the numbers in TREE
  '''

  # base case, when there is just 1 node, so just return a tuple of that
  if type(TREE) is not tuple:
    return tuple([TREE])
  # when TREE is more than just a node, traverse the tree
  else:
    return TREE_ORDER(TREE[0]) + TREE_ORDER(TREE[2]) + TREE_ORDER(TREE[1])
```

```python
TREE_ORDER(42)
```

```python
TREE_ORDER(((1, 2, 3), 7, 8))
```

```python
TREE_ORDER(((3, 7, 10), 15, ((16, 18, 20), 30, 100)))
```
