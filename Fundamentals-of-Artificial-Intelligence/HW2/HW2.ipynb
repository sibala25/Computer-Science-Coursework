```python
# Question 1
def BFS(TREE):
  '''
  A function which given a tree, returns a tuple of leaf nodes in the order
  they would be visited in a left-to-right breadth-first search.

  Arguments:
  - TREE: tree to traverse

  Returns:
  - tuple, order of leaf nodes visited in a left-to-right breadth-first search
  '''
  # base case if tree is a single leaf, only one order
  # return it as a tuple
  if type(TREE) != tuple:
    return tuple([TREE])
  # if not at a leaf, use a list as a queue and add nodes from left to right
  else:
    queue = [TREE]
    bfs = []

    while len(queue) > 0:
      node = queue.pop(0)
      # if not at a leaf node yet
      if type(node) == tuple:
          # add children left to right
          for child in node:
            queue.append(child)
      # reached a leaf node, so append to output
      else:
        bfs.append(node)

    return tuple(bfs)
```

```python
# Question 2
def DFS(TREE):
  '''
  A function which given a tree, returns a tuple of leaf nodes in the order
  they would be visited in a left-to-right depth-first search.

  Arguments:
  - TREE: tree to traverse

  Returns:
  - tuple, order of leaf nodes visited in a left-to-right depth-first search
  '''
  # base case if tree is a single leaf, only one order
  # return it as a tuple
  if type(TREE) != tuple:
    return tuple([TREE])
  # if not at a leaf, use a list as a stack and add nodes
  else:
    stack = [TREE]
    dfs = []

    while len(stack) > 0:
      # pop from end (stack behavior)
      node = stack.pop()
      # if not at a leaf node yet
      if type(node) == tuple:
          # add children right to left
          for child in node[::-1]:
            stack.append(child)
      # are at a leaf node, so append to output
      else:
        dfs.append(node)

    return tuple(dfs)
```

```python
# Question 3
def DLDFS(TREE,D,L):
  '''
  A function which given a tree, current depth, and a depth limit, returns a tuple
  of leaf nodes in the order that they would be visited in a right-to-left
  depth-first search, until the depth limit.

  Arguments:
  - TREE: tree to traverse
  - D: current depth in search
  - L: depth limit

  Returns:
  - tuple, order of leaf nodes visited in a right-to-left depth-first search
  '''

  # base case if tree is a single leaf, only one order, return as tuple
  if type(TREE) != tuple:
    return tuple([TREE])

  # if not at a leaf, use a list as a stack and add nodes
  # ensure you do not surpass the depth limit - track depth
  else:
    stack = [(TREE,D)]
    dfs = []

    while len(stack) > 0:
      # pop from end
      node, depth = stack.pop()
      # if not at a leaf node yet and depth limit not reached
      if type(node) == tuple:
        # and if we didn't pass the limit
        if depth < L:
            # add children left to right (so we visit right to left)
            for child in node:
              stack.append((child, depth + 1))
      # are at a leaf node, so append to output
      else:
        dfs.append(node)

    return tuple(dfs)

def DFID(TREE,D):
  '''
  A function which given a tree and its depth, returns a tuple of leaf nodes in
  the order that they would be visited in a right-to-left depth-first
  iterative-deepening search.

  Arguments:
  - TREE: tree to traverse
  - D: depth of tree

  Returns:
  - tuple, order of leaf nodes visited in a right-to-left depth-first iterative
  deepening search
  '''
  result = []

  # run depth-limited DFS for all possible depth limits from 0 to D
  for i in range(D+1):
    # add to result
    result.extend(DLDFS(TREE,0,i))
  return tuple(result)
```

```python
# Question 4

# These functions implement a depth-first solver for the homer-baby-dog-poison
# problem. In this implementation, a state is represented by a single tuple
# (homer, baby, dog, poison), where each variable is True if the respective entity is
# on the west side of the river, and False if it is on the east side.
# Thus, the initial state for this problem is (False False False False) (everybody
# is on the east side) and the goal state is (True True True True).

# The main entry point for this solver is the function DFS_SOL, which is called
# with (a) the state to search from and (b) the path to this state. It returns
# the complete path from the initial state to the goal state: this path is a
# list of intermediate problem states. The first element of the path is the
# initial state and the last element is the goal state. Each intermediate state
# is the state that results from applying the appropriate operator to the
# preceding state. If there is no solution, DFS_SOL returns [].
# To call DFS_SOL to solve the original problem, one would call
# DFS_SOL((False, False, False, False), [])
# However, it should be possible to call DFS_SOL with any intermediate state (S)
# and the path from the initial state to S (PATH).

# First, we define the helper functions of DFS_SOL.

# FINAL_STATE takes a single argument S, the current state, and returns True if it
# is the goal state (True, True, True, True) and False otherwise.
def FINAL_STATE(S):
    # true if all characters on goal side, false otherwise
    return S == (True, True, True, True)


# NEXT_STATE returns the state that results from applying an operator to the
# current state. It takes two arguments: the current state (S), and which entity
# to move (A, equal to "h" for homer only, "b" for homer with baby, "d" for homer
# with dog, and "p" for homer with poison).
# It returns a list containing the state that results from that move.
# If applying this operator results in an invalid state (because the dog and baby,
# or poisoin and baby are left unsupervised on one side of the river), or when the
# action is impossible (homer is not on the same side as the entity) it returns [].
# NOTE that NEXT_STATE returns a list containing the successor state (which is
# itself a tuple)# the return should look something like [(False, False, True, True)].
def NEXT_STATE(S, A):
    # extract current locations
    h, b, d, p = S
    # tuple to return with new locations
    move = None

    # first make the move

    if A == "h":
      # move Homer to other side
      move = (not h, b, d, p)
    # if baby and Homer are on same side, move them
    elif A == "b" and h == b:
      move = (not h, not b, d, p)
    # if dog and Homer are on same side, move them
    elif A == "d" and h == d:
      move = (not h, b, not d, p)
    # if poison and Homer on same side, move them
    elif A == "p" and h == p:
      move = (not h, b, d, not p)
    # invalid move or not on same side
    else:
      return []

    # then, check that constraints are not violated
    h, b, d, p = move
    # baby and poison are unsupervised
    if h != b and b == p:
      return []
    # baby and dog are unsupervised
    if h !=  b and b == d:
      return []

    return [move]

# SUCC_FN returns all of the possible legal successor states to the current
# state. It takes a single argument (S), which encodes the current state, and
# returns a list of each state that can be reached by applying legal operators
# to the current state.
def SUCC_FN(S):
    successors = []
    # find all successor states for any move using NEXT_STATE
    for A in ["h","b","d","p"]:
      result = NEXT_STATE(S,A)
      # add only legal states
      if result != []:
        successors.extend(result)
    return successors

# ON_PATH checks whether the current state is on the stack of states visited by
# this depth-first search. It takes two arguments: the current state (S) and the
# stack of states visited by DFS (STATES). It returns True if S is a member of
# STATES and False otherwise.
def ON_PATH(S, STATES):
    if S in STATES:
      return True           # S has been visited
    else:
      return False          # S has not been visited

# MULT_DFS is a helper function for DFS_SOL. It takes two arguments: a list of
# states from the initial state to the current state (PATH), and the legal
# successor states to the last, current state in the PATH (STATES). PATH is a
# first-in first-out list of states# that is, the first element is the initial
# state for the current search and the last element is the most recent state
# explored. MULT_DFS does a depth-first search on each element of STATES in
# turn. If any of those searches reaches the final state, MULT_DFS returns the
# complete path from the initial state to the goal state. Otherwise, it returns
# [].
def MULT_DFS(STATES, PATH):
    # explore each successor state
    for S in STATES:
      # if goal is reached, we have our complete path
      if FINAL_STATE(S):
        return PATH + [S]
      # if state has already been visited, skip it
      if ON_PATH(S, PATH):
        continue
      else:
        # recursively search from this state with updated path
        result = MULT_DFS(SUCC_FN(S), PATH + [S])
        # goal state path found
        if result != []:
          return result
    return []


# DFS_SOL does a depth first search from a given state to the goal state. It
# takes two arguments: a state (S) and the path from the initial state to S
# (PATH). If S is the initial state in our search, PATH is set to []. DFS_SOL
# performs a depth-first search starting at the given state. It returns the path
# from the initial state to the goal state, if any, or [] otherwise. DFS_SOL is
# responsible for checking if S is already the goal state, as well as for
# ensuring that the depth-first search does not revisit a node already on the
# search path (i.e., S is not on PATH).
def DFS_SOL(S, PATH):
     # check if S is goal state
     if FINAL_STATE(S):
       return PATH + [S]

     # check if S has already been visited in the search path
     if ON_PATH(S,PATH):
       return []

     # continue DFS from succesor states
     return MULT_DFS(SUCC_FN(S), PATH + [S])
```
